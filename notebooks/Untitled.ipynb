```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path("../data/raw/2018")

file_mapping = {
    "pressure": "2018_SCADA_Pressures.csv",
    "flow": "2018_SCADA_Flows.csv",
    "demand": "2018_SCADA_Demands.csv",
    "leakage": "2018_Leakages.csv"
}

# Load all datasets with correct semicolon and decimal settings
datasets = {
    name: pd.read_csv(DATA_DIR / filename, sep=";", decimal=",") 
    for name, filename in file_mapping.items()
}

# Access individual DataFrames like this:
pressure = datasets["pressure"]
flow = datasets["flow"]
demand = datasets["demand"]
leakage = datasets["leakage"]

for name, df in datasets.items():
    print(f"\n{name.upper()}")
    print("Shape:", df.shape)
    print("Columns:", df.columns.tolist()[:10])
    print(df.head(3))
    print("Missing values:", df.isna().sum().sum())
```
