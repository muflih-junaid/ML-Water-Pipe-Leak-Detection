```python
import pandas as pd
import numpy as np

# Load datasets
X_train = pd.read_csv("../data/processed/X_train.csv")
X_val = pd.read_csv("../data/processed/X_val.csv")

y_train = pd.read_csv("../data/processed/y_train.csv")
y_val = pd.read_csv("../data/processed/y_val.csv")

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
```

```python
X_train.head()
```

```python
print(X_train.isnull().sum())
print(X_val.isnull().sum())
```

```python
print(y_train.isnull().sum())
print(y_val.isnull().sum())
```

```python
print("Duplicates:", X_train.duplicated().sum())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)
```

```python
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)
```

```python
print(X_train_scaled.head())
```

```python
print("Training labels:")
print(y_train.value_counts())

print("Validation labels:")
print(y_val.value_counts())
```

```python
import os
import pandas as pd

# Export preprocessed datasets
X_train_scaled.to_csv(
    "../data/processed/X_train_processed.csv", index=False
)

X_val_scaled.to_csv(
    "../data/processed/X_val_processed.csv", index=False
)

y_train.to_csv(
    "../data/processed/y_train_processed.csv", index=False
)

y_val.to_csv(
    "../data/processed/y_val_processed.csv", index=False
)

print("All preprocessed datasets exported successfully!")
```
